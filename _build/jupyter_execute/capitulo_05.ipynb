## 5.1 Bucle `while`
El bucle `while` repite un bloque de código **mientras** una condición sea verdadera.

Sintaxis:
```python
while condicion:
    bloque
```

Es ideal cuando no sabemos cuántas repeticiones habrá, pero sí conocemos la condición de corte.


````python
contador = 1
while contador <= 5:
    print("Iteración", contador)
    contador += 1
````

### Ejemplo: pedir números hasta ingresar 0

## 5.2 Bucle `for`
El bucle `for` recorre secuencias: listas, cadenas, tuplas, diccionarios o rangos.

Sintaxis:
```python
for variable in secuencia:
    bloque
```

Es ideal cuando sabemos cuántas iteraciones habrá.


````python
for letra in "UNO":
    print(letra)
````

### Recorrer una lista


````python
numeros = [10, 20, 30]
for n in numeros:
    print(n)
````

## 5.3 Uso de `range()`
`range()` genera secuencias de números.

Formas:
- `range(fin)`
- `range(inicio, fin)`
- `range(inicio, fin, paso)`


````python
list(range(5)), list(range(2, 6)), list(range(0, 10, 2))
````

### Ejemplo: imprimir números del 1 al 5


````python
for i in range(1, 6):
    print(i)
````

## 5.4 Control de bucles
Python ofrece tres instrucciones especiales para controlar bucles:

- `break`: interrumpe el bucle
- `continue`: salta a la siguiente iteración
- `pass`: no hace nada (placeholder)


### 5.4.1 `break`
Detiene el bucle inmediatamente.


````python
for i in range(10):
    if i == 4:
        break
    print(i)
````

### 5.4.2 `continue`
Salta la iteración actual.


````python
for i in range(6):
    if i % 2 == 0:
        continue
    print(i)
````

### 5.4.3 `pass`
Se usa cuando necesitamos un bloque vacío.


````python
for i in range(3):
    pass
print("Bucle ejecutado con pass")
````

## 5.5 Bucles anidados
Un bucle dentro de otro permite recorrer estructuras más complejas.

Ejemplo: tabla de multiplicar.


````python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, "x", j, "=", i*j)
    print("---")
````

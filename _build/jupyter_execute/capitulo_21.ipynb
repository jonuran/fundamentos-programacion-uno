La **notación Big O** es una herramienta matemática que permite describir el rendimiento de un algoritmo en función del tamaño de entrada `n`.

Big O responde preguntas como:
- ¿Cómo escala un algoritmo cuando crecen los datos?
- ¿Cuál es el peor caso posible?
- ¿Qué tan eficiente es un algoritmo comparado con otro?

En este capítulo veremos las clases de complejidad más comunes:
- 21.1 O(1)
- 21.2 O(log n)
- 21.3 O(n)
- 21.4 O(n log n)
- 21.5 O(n²)
- 21.6 O(2ⁿ)

Estas clases son fundamentales para evaluar algoritmos y estructuras de datos.

## 21.1 O(1) — Tiempo constante
Un algoritmo es **O(1)** si su tiempo de ejecución no depende del tamaño de la entrada.

Ejemplos típicos:
- acceso a un elemento de una lista por índice
- acceso a un valor en un diccionario
- operaciones aritméticas

Ejemplo:

````python
def acceso_constante(lista):
    return lista[0]  # siempre tarda lo mismo

acceso_constante([10, 20, 30])
````

O(1) es la complejidad ideal: no importa cuán grande sea la entrada, el tiempo no cambia.

## 21.2 O(log n) — Logarítmica
Los algoritmos **logarítmicos** son extremadamente eficientes.

Aparecen cuando el problema se reduce a la mitad en cada paso.

Ejemplos:
- búsqueda binaria
- árboles balanceados (AVL, Red-Black)
- heaps

### Ejemplo: búsqueda binaria

````python
def busqueda_binaria(lista, objetivo):
    izq, der = 0, len(lista) - 1
    while izq <= der:
        medio = (izq + der) // 2
        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            izq = medio + 1
        else:
            der = medio - 1
    return -1

busqueda_binaria([1,2,3,4,5,6,7,8], 6)
````

Cada paso reduce el espacio de búsqueda a la mitad → **O(log n)**.

Es una de las complejidades más deseables.

## 21.3 O(n) — Lineal
Un algoritmo es **O(n)** si su tiempo crece proporcionalmente al tamaño de la entrada.

Ejemplos:
- recorrer una lista
- buscar un elemento sin ordenar
- sumar todos los elementos

Ejemplo:

````python
def suma_lineal(lista):
    total = 0
    for x in lista:
        total += x
    return total

suma_lineal([1,2,3,4,5])
````

O(n) es eficiente para listas pequeñas o medianas, pero puede ser costoso para millones de elementos.

## 21.4 O(n log n) — Lineal-logarítmica
Esta complejidad aparece en algoritmos que combinan:
- un proceso lineal
- repetido logarítmicamente

Ejemplos:
- algoritmos de ordenamiento eficientes (MergeSort, QuickSort)
- algunos algoritmos de grafos

### Ejemplo conceptual
MergeSort divide la lista en mitades (log n) y luego combina cada mitad (n).

````python
def merge_sort(lista):
    if len(lista) <= 1:
        return lista
    medio = len(lista) // 2
    izq = merge_sort(lista[:medio])
    der = merge_sort(lista[medio:])

    resultado = []
    i = j = 0
    while i < len(izq) and j < len(der):
        if izq[i] < der[j]:
            resultado.append(izq[i])
            i += 1
        else:
            resultado.append(der[j])
            j += 1
    resultado.extend(izq[i:])
    resultado.extend(der[j:])
    return resultado

merge_sort([5,3,8,1,2])
````

O(n log n) es la complejidad óptima para ordenamientos basados en comparaciones.

## 21.5 O(n²) — Cuadrática
Los algoritmos cuadráticos aparecen cuando hay **dos bucles anidados**.

Ejemplos:
- Bubble Sort
- Selection Sort
- comparaciones entre todos los pares

### Ejemplo clásico: Bubble Sort

````python
def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n-i-1):
            if lista[j] > lista[j+1]:
                lista[j], lista[j+1] = lista[j+1], lista[j]
    return lista

bubble_sort([5,3,8,1,2])
````

O(n²) es aceptable para n pequeño, pero se vuelve muy lento para n grande.

Ejemplo visual (crecimiento):
```
n = 10 → 100 operaciones
n = 100 → 10.000 operaciones
n = 1000 → 1.000.000 operaciones
```

## 21.6 O(2ⁿ) — Exponencial
Los algoritmos exponenciales crecen **explosivamente**.

Ejemplos:
- Fibonacci recursivo
- problemas de fuerza bruta
- subsets, combinaciones, permutaciones
- muchos problemas NP-completos

### Ejemplo: Fibonacci recursivo (sin optimización)

````python
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)

fib(10)
````

Crecimiento exponencial:
```
n = 10 → 100 llamadas
n = 20 → 10.000 llamadas
n = 30 → 1.000.000 llamadas
```

Los algoritmos exponenciales son imprácticos para valores grandes de `n`.

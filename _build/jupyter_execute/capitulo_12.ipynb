## 12.1 Apertura de archivos
Para trabajar con un archivo, primero debemos **abrirlo**.

La función principal es:
```python
open(ruta, modo)
```

Donde:
- **ruta**: ubicación del archivo (puede ser relativa o absoluta)
- **modo**: indica si queremos leer, escribir, agregar, etc.

Ejemplo básico:

````python
archivo = open("ejemplo.txt", "w")
archivo.write("Hola mundo")
archivo.close()
````

La apertura de archivos puede fallar si:
- el archivo no existe (modo lectura)
- no tenemos permisos
- la ruta es incorrecta

Por eso es recomendable usar `try-except` o `with`.

## 12.2 Escritura
Para escribir en un archivo usamos el modo `w` o `a`.

### Modo `w` — sobrescribe el archivo
Si el archivo existe, se borra su contenido.

Ejemplo:

````python
with open("datos.txt", "w") as f:
    f.write("Primera línea\n")
    f.write("Segunda línea\n")
````

### Modo `a` — agregar al final
No borra el contenido previo.

````python
with open("datos.txt", "a") as f:
    f.write("Línea agregada al final\n")
````

## 12.3 Lectura
Python ofrece varios métodos para leer archivos:

- `read()` → lee todo el archivo como un string
- `readline()` → lee una sola línea
- `readlines()` → devuelve una lista de líneas
- iterar directamente sobre el archivo

Ejemplo con `read()`:

````python
with open("datos.txt", "r") as f:
    contenido = f.read()
contenido
````

Lectura línea por línea (ideal para archivos grandes):

````python
with open("datos.txt", "r") as f:
    for linea in f:
        print(linea.strip())
````

## 12.4 Cierre
Cuando abrimos un archivo con `open()`, debemos cerrarlo con `close()`.

Sin embargo, la forma recomendada es usar `with`, que cierra el archivo automáticamente incluso si ocurre un error.

Ejemplo incorrecto (puede dejar archivos abiertos):
```python
f = open("archivo.txt")
contenido = f.read()
# falta f.close()
```

Ejemplo correcto:

````python
with open("archivo.txt", "w") as f:
    f.write("Texto seguro")
````

## 12.5 Modos de apertura
Los modos más comunes son:

| Modo | Descripción |
|------|-------------|
| `r` | Leer (error si no existe) |
| `w` | Escribir (sobrescribe) |
| `a` | Agregar al final |
| `r+` | Leer y escribir |
| `b` | Modo binario |
| `rb` | Leer binario |
| `wb` | Escribir binario |

Ejemplo de lectura binaria (imágenes, PDFs, etc.):

````python
with open("imagen.png", "rb") as f:
    datos = f.read()
````

## 12.6 Manipulación de archivos
Python ofrece el módulo `os` para trabajar con archivos y rutas.

### Verificar si un archivo existe

````python
import os
os.path.exists("datos.txt")
````

### Renombrar archivos

````python
os.rename("datos.txt", "datos_renombrado.txt")
````

### Eliminar archivos

````python
os.remove("datos_renombrado.txt")
````

## 12.7 Manejo de directorios
El módulo `os` también permite trabajar con carpetas.

### Crear un directorio

````python
os.mkdir("carpeta_ejemplo")
````

### Listar archivos de un directorio

````python
os.listdir(".")
````

### Cambiar de directorio

````python
os.chdir("carpeta_ejemplo")
os.getcwd()
````

### Volver al directorio anterior

````python
os.chdir("..")
os.getcwd()
````

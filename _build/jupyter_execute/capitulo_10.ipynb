## 10.1 Definición de funciones
Una función se define con la palabra clave `def`, seguida del nombre y paréntesis.

Sintaxis general:
```python
def nombre_funcion(parámetros):
    instrucciones
```

Ejemplo básico:

````python
def saludar():
    print("Hola, bienvenido al curso de Python")

saludar()
````

Una función puede estar vacía usando `pass`:

````python
def futura_funcion():
    pass
````

## 10.2 Parámetros y argumentos
Los **parámetros** son variables definidas en la función.
Los **argumentos** son los valores reales que se pasan al llamar la función.

Ejemplo:

````python
def saludar(nombre):
    print(f"Hola {nombre}")

saludar("Jonatan")
````

Python permite **argumentos posicionales** y **argumentos nombrados**:

````python
def presentar(nombre, edad):
    print(f"Nombre: {nombre}, Edad: {edad}")

presentar("Ana", 22)  # posicional
presentar(edad=30, nombre="Luis")  # nombrado
````

## 10.3 Retorno de valores
Las funciones pueden devolver valores usando `return`.

Ejemplo:

````python
def sumar(a, b):
    return a + b

resultado = sumar(10, 5)
resultado
````

Una función puede retornar múltiples valores (como tupla implícita):

````python
def operaciones(a, b):
    return a + b, a - b, a * b

suma, resta, producto = operaciones(10, 3)
suma, resta, producto
````

Si no se usa `return`, la función devuelve `None`:

````python
def ejemplo():
    print("Hola")

valor = ejemplo()
valor
````

## 10.4 Argumentos por defecto
Los parámetros pueden tener valores por defecto.

Esto permite llamar la función sin pasar todos los argumentos.

Ejemplo:

````python
def saludar(nombre="invitado", mensaje="Bienvenido"):
    print(f"Hola {nombre}, {mensaje}")

saludar()
saludar("Ana")
saludar("Luis", "¿Cómo estás?")
````

⚠️ **Importante:** Los valores por defecto se evalúan una sola vez.

Esto puede causar comportamientos inesperados con listas o diccionarios mutables.

````python
def agregar(elemento, lista=[]):
    lista.append(elemento)
    return lista

agregar(1), agregar(2), agregar(3)
````

Solución correcta:

````python
def agregar_correcto(elemento, lista=None):
    if lista is None:
        lista = []
    lista.append(elemento)
    return lista

agregar_correcto(1), agregar_correcto(2), agregar_correcto(3)
````

## 10.5 Argumentos variables (`*args`, `**kwargs`)
### `*args` — argumentos posicionales variables
Permite recibir cualquier cantidad de argumentos.

### `**kwargs` — argumentos nombrados variables
Permite recibir pares clave-valor arbitrarios.

Ejemplos:

````python
def sumar_todo(*args):
    return sum(args)

sumar_todo(1, 2, 3, 4, 5)
````

````python
def mostrar_info(**kwargs):
    return kwargs

mostrar_info(nombre="Jon", edad=39, ciudad="Buenos Aires")
````

Incluso pueden combinarse:

````python
def mezcla(a, *args, **kwargs):
    return a, args, kwargs

mezcla(1, 2, 3, x=10, y=20)
````

## 10.6 Ámbito de variables (scope)
El **scope** determina dónde es accesible una variable.

### Tipos de ámbito:
- **Local**: dentro de una función.
- **Global**: fuera de funciones.
- **Nonlocal**: en funciones anidadas.

Ejemplo de variable local vs global:

````python
x = 10  # global

def prueba():
    x = 5  # local
    return x

prueba(), x
````

### Uso de `global` (no recomendado salvo casos especiales)

````python
contador = 0

def incrementar():
    global contador
    contador += 1

incrementar()
contador
````

### Uso de `nonlocal` en funciones anidadas

````python
def externa():
    x = 10
    def interna():
        nonlocal x
        x = 20
    interna()
    return x

externa()
````

## 10.7 Paso por referencia
Python usa **paso por asignación**.

Esto significa:
- Los objetos mutables pueden modificarse dentro de la función.
- Los objetos inmutables no pueden modificarse.

Ejemplo con lista (mutable):

````python
def agregar(lista):
    lista.append(100)

numeros = [1, 2, 3]
agregar(numeros)
numeros
````

Ejemplo con entero (inmutable):

````python
def cambiar(n):
    n = 99

x = 10
cambiar(x)
x
````

## 10.8 Documentación de funciones (docstrings)
Los **docstrings** permiten documentar funciones de forma estándar.

Se escriben entre triple comillas justo después de la definición.

Ejemplo:

````python
def sumar(a, b):
    """
    Devuelve la suma de dos números.

    Parámetros:
        a (int/float): primer número
        b (int/float): segundo número

    Retorna:
        int/float: resultado de la suma
    """
    return a + b

help(sumar)
````

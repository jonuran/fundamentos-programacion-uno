## 3.1 Operadores aritméticos
Los operadores aritméticos permiten realizar cálculos matemáticos. Python respeta las reglas matemáticas tradicionales.

Operadores disponibles:
- `+` suma
- `-` resta
- `*` multiplicación
- `/` división
- `//` división entera
- `%` módulo (resto)
- `**` potencia

Ejemplos básicos:


```python
5 + 3, 5 - 3, 5 * 3, 5 / 3
```

```python
5 // 3, 5 % 3, 2 ** 4
```

### División entera y módulo
- `//` devuelve solo la parte entera
- `%` devuelve el resto

Son muy útiles en algoritmos, por ejemplo para determinar si un número es par:


```python
numero = 14
es_par = (numero % 2 == 0)
es_par
```

## 3.2 Operadores lógicos
Los operadores lógicos permiten combinar condiciones.

Operadores:
- `not`: negación
- `and`: conjunción
- `or`: disyunción

Estos operadores trabajan con valores booleanos (`True`, `False`).

Ejemplos:


```python
not True, True and False, True or False
```

### Tabla de verdad
| x | y | not x | x and y | x or y |
|---|---|-------|---------|--------|
| False | False | True | False | False |
| False | True | True | False | True |
| True | False | False | False | True |
| True | True | False | True | True |


## 3.3 Operadores de comparación
Permiten comparar valores y devuelven un booleano.

Operadores:
- `==` igual que
- `!=` distinto que
- `<` menor que
- `>` mayor que
- `<=` menor o igual que
- `>=` mayor o igual que

Ejemplos:


```python
3 == 3, 3 != 4, 3 < 5, 3 >= 3
```

### Comparación de cadenas
Python compara cadenas según el orden ASCII.

Ejemplo:
- `'a' > 'A'` es `True` porque las minúsculas tienen mayor valor ASCII.


```python
'a' > 'A', 'Python' == 'python'
```

## 3.4 Prioridad de operadores
Python evalúa las expresiones siguiendo un orden específico:

1. Paréntesis `()`
2. Potencias `**`
3. Multiplicación, división, módulo, división entera
4. Suma y resta
5. Comparaciones
6. `not`
7. `and`
8. `or`

Ejemplo complejo:


```python
resultado = 2 + 3 * 4 > 10 and not False
resultado
```

Desglose:
- `3 * 4` → 12
- `2 + 12` → 14
- `14 > 10` → True
- `not False` → True
- `True and True` → True


## 3.5 Expresiones
Una expresión combina valores, variables y operadores para producir un resultado.

Ejemplo clásico:


```python
a = 5
b = 3
c = (a + b) * 2
c
```

## 22.1 Regla de la suma
La **regla de la suma** se aplica cuando un algoritmo realiza **dos tareas consecutivas**, una después de la otra.

Si un algoritmo realiza:
- una operación de complejidad **O(f(n))**, y luego
- otra operación de complejidad **O(g(n))**,

entonces la complejidad total es:

### ✔ O(f(n) + g(n))

### Ejemplo 1
```python
for i in range(n):      # O(n)
    ...

for j in range(n):      # O(n)
    ...
```
Complejidad total:

→ O(n + n) = **O(n)**

### Ejemplo 2: diferentes tamaños
```python
for i in range(n):      # O(n)
    ...

for j in range(n*n):    # O(n²)
    ...
```
Complejidad total:

→ O(n + n²) = **O(n²)** (domina el término mayor)

### Ejemplo práctico en Python

````python
def regla_suma(n):
    for _ in range(n):
        pass  # O(n)
    for _ in range(n*n):
        pass  # O(n²)
    return "Completado"

regla_suma(10)
````

La regla de la suma nos permite analizar algoritmos compuestos por partes independientes.

## 22.2 Regla de la multiplicación
La **regla de la multiplicación** se aplica cuando un algoritmo realiza una tarea **dentro de otra**.

Si un algoritmo realiza:
- una operación de complejidad **O(f(n))**, y dentro de ella
- otra operación de complejidad **O(g(n))**,

entonces la complejidad total es:

### ✔ O(f(n) × g(n))

### Ejemplo 1: doble bucle
```python
for i in range(n):      # O(n)
    for j in range(n):  # O(n)
        ...
```
Complejidad total:

→ O(n × n) = **O(n²)**

### Ejemplo 2: bucles de distinto tamaño
```python
for i in range(n):      # O(n)
    for j in range(n*n):  # O(n²)
        ...
```
Complejidad total:

→ O(n × n²) = **O(n³)**

### Ejemplo práctico en Python

````python
def regla_multiplicacion(n):
    for _ in range(n):
        for _ in range(n):
            pass  # O(n²)
    return "Completado"

regla_multiplicacion(10)
````

La regla de la multiplicación es clave para analizar algoritmos con bucles anidados o estructuras jerárquicas.

## 22.3 Comparación de algoritmos
Para elegir el mejor algoritmo, debemos comparar sus complejidades.

### Orden de eficiencia (de mejor a peor)
```
O(1)
O(log n)
O(n)
O(n log n)
O(n²)
O(2ⁿ)
O(n!)
```

### Ejemplo comparativo
Supongamos que queremos buscar un elemento en una lista:

- **Búsqueda lineal** → O(n)
- **Búsqueda binaria** → O(log n)

Claramente, la búsqueda binaria es mucho más eficiente.

### Ejemplo práctico: comparación de tiempos

````python
import time

def busqueda_lineal(lista, objetivo):
    for i, x in enumerate(lista):
        if x == objetivo:
            return i
    return -1

def busqueda_binaria(lista, objetivo):
    izq, der = 0, len(lista) - 1
    while izq <= der:
        medio = (izq + der) // 2
        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            izq = medio + 1
        else:
            der = medio - 1
    return -1

lista = list(range(1_000_000))

inicio = time.perf_counter()
busqueda_lineal(lista, 999_999)
print("Lineal:", time.perf_counter() - inicio)

inicio = time.perf_counter()
busqueda_binaria(lista, 999_999)
print("Binaria:", time.perf_counter() - inicio)
````

La diferencia de rendimiento es enorme, especialmente para valores grandes de `n`.

Comparar algoritmos es esencial para elegir la solución más eficiente.

## 22.4 Optimización
La optimización consiste en mejorar el rendimiento de un algoritmo.

### Estrategias comunes
#### ✔ 1. Evitar trabajo innecesario
Ejemplo: cortar un bucle cuando ya encontramos lo que buscamos.

#### ✔ 2. Usar estructuras de datos eficientes
- listas → O(n)
- diccionarios → O(1)
- heaps → O(log n)

#### ✔ 3. Reducir bucles anidados
Pasar de O(n²) a O(n log n) o O(n).

#### ✔ 4. Usar algoritmos más eficientes
Ejemplo: reemplazar Bubble Sort (O(n²)) por MergeSort (O(n log n)).

#### ✔ 5. Memoización
Guardar resultados ya calculados.

### Ejemplo de optimización: Fibonacci
Versión lenta (O(2ⁿ)):

````python
def fib_lento(n):
    if n <= 1:
        return n
    return fib_lento(n-1) + fib_lento(n-2)

fib_lento(10)
````

Versión optimizada con memoización (O(n)):

````python
memo = {}

def fib_rapido(n):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib_rapido(n-1) + fib_rapido(n-2)
    return memo[n]

fib_rapido(30)
````

La optimización puede transformar un algoritmo inutilizable en uno eficiente.

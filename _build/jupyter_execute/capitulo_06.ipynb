## 6.1 Definición de cadenas
Una cadena es una secuencia ordenada de caracteres. Puede incluir letras, números, símbolos y espacios.

Se puede crear usando comillas simples o dobles:
```python
cadena1 = "Hola UNO"
cadena2 = 'Python es genial'
```

Python también permite cadenas multilínea usando triple comilla:
```python
texto = """
Esto es una cadena
de varias líneas.
"""
```


````python
cadena = "Hola UNO"
cadena
````

## 6.2 Indexación
Cada carácter de una cadena tiene una posición (índice), comenzando desde 0.

Ejemplo:
```
H  o  l  a     U  N  O
0  1  2  3  4  5  6  7
```

Para acceder a un carácter usamos corchetes:


````python
texto = "Hola UNO"
texto[0], texto[4], texto[7]
````

## 6.3 Indexación negativa
Python permite acceder a los caracteres desde el final usando índices negativos.

```
H  o  l  a     U  N  O
-8 -7 -6 -5 -4 -3 -2 -1
```

Ejemplo:


````python
texto = "Hola UNO"
texto[-1], texto[-3], texto[-8]
````

## 6.4 Subcadenas (slicing)
El slicing permite obtener partes de una cadena.

Sintaxis:
```
cadena[inicio:fin]
```

Python incluye el índice `inicio` pero excluye `fin`.

Ejemplos:


````python
texto = "Hola UNO"
texto[0:4], texto[5:8], texto[:4], texto[5:], texto[:]
````

## 6.5 Operaciones con cadenas
Las cadenas permiten varias operaciones útiles.

### Concatenación
Se usa el operador `+`.


````python
"Hola " + "UNO"
````

### Repetición
Se usa el operador `*`.


````python
"Python " * 3
````

### Pertenencia (`in`)
Permite verificar si una subcadena está dentro de otra.


````python
"UNO" in "Hola UNO"
````

## 6.6 Comparación de cadenas
Las cadenas se comparan según el orden ASCII.

Ejemplos:
- `'a' > 'A'` es verdadero
- `'hola' == 'hola'` es verdadero
- `'uno' < 'dos'` depende del orden alfabético


````python
'a' > 'A', 'hola' == 'hola', 'uno' < 'dos'
````

## 6.7 Funciones de cadenas
Python incluye funciones y métodos muy útiles para trabajar con texto.

### 6.7.1 `len()`
Devuelve la longitud de la cadena.


````python
len("Hola UNO")
````

### 6.7.2 `upper()` y `lower()`
Transforman la cadena a mayúsculas o minúsculas.


````python
"Hola UNO".upper(), "Hola UNO".lower()
````

### 6.7.3 `split()`
Divide la cadena en una lista usando un separador.


````python
"Python es genial".split()
````

### 6.7.4 `replace()`
Reemplaza una subcadena por otra.


````python
"Hola UNO".replace("UNO", "Python")
````

## 6.8 Formateo de cadenas
Python ofrece varias formas de insertar valores dentro de una cadena.

### 6.8.1 `format()`
Permite reemplazar llaves `{}` por valores.


````python
"Hola {}".format("UNO")
````

### 6.8.2 f-strings
Son la forma más moderna y recomendada.

Se escriben anteponiendo `f` a la cadena.


````python
nombre = "UNO"
f"Hola {nombre}"
````

## 17.1 Arreglos (listas)
En Python, los arreglos se implementan como **listas dinámicas**.

Características:
- Tamaño dinámico (crecen y se reducen automáticamente)
- Acceso por índice en tiempo O(1)
- Inserciones y eliminaciones en el medio → O(n)
- Permiten elementos heterogéneos

Ejemplo básico:

```python
lista = [10, 20, 30, 40]
lista, lista[2]
```

### Operaciones comunes
- `append()` → agrega al final
- `insert()` → inserta en posición
- `pop()` → elimina y devuelve un elemento
- `remove()` → elimina por valor
- `sort()` → ordena
- `reverse()` → invierte

Ejemplo:

```python
lista.append(50)
lista.insert(1, 15)
lista.pop()
lista
```

Las listas son la estructura más flexible de Python, pero no siempre la más eficiente para todas las operaciones.

## 17.2 Pilas (LIFO)
Una **pila** es una estructura donde el último elemento en entrar es el primero en salir.

LIFO → *Last In, First Out*

Ejemplos de uso:
- Deshacer acciones (Ctrl+Z)
- Recursión
- Evaluación de expresiones
- Navegación en el historial del navegador

En Python, una pila puede implementarse con una lista usando `append()` y `pop()`.

```python
pila = []
pila.append(1)
pila.append(2)
pila.append(3)
pila, pila.pop(), pila
```

Complejidad:
- `push` → O(1)
- `pop` → O(1)

Las pilas son extremadamente eficientes y simples.

## 17.3 Colas (FIFO)
Una **cola** es una estructura donde el primer elemento en entrar es el primero en salir.

FIFO → *First In, First Out*

Ejemplos de uso:
- Procesos en un sistema operativo
- Turnos en un banco
- Simulaciones
- Sistemas de mensajería

Implementación básica con listas (no eficiente):

```python
cola = []
cola.append("A")
cola.append("B")
cola.append("C")
cola.pop(0), cola
```

Eliminar del inicio de una lista es O(n), por lo que no es eficiente.

La forma correcta es usar `deque`.

## 17.4 Deque
`deque` (double-ended queue) es una cola de doble extremo.

Ventajas:
- Inserciones y eliminaciones en ambos extremos → O(1)
- Ideal para colas y pilas
- Muy eficiente

Se importa desde `collections`.

```python
from collections import deque

dq = deque([1, 2, 3])
dq.append(4)
dq.appendleft(0)
dq.pop(), dq.popleft(), dq
```

`deque` es una de las estructuras más eficientes de Python para manejo de secuencias dinámicas.

## 17.5 Tabla hash
Una **tabla hash** es una estructura que permite almacenar pares clave-valor con acceso en tiempo O(1) promedio.

En Python, las tablas hash se implementan como **diccionarios (`dict`)**.

Características:
- Acceso rápido por clave
- No tienen orden (hasta Python 3.6, ahora mantienen orden de inserción)
- Claves deben ser inmutables (strings, números, tuplas)

Ejemplo básico:

```python
tabla = {"nombre": "Ana", "edad": 22, "nota": 9}
tabla["nombre"], tabla.get("nota")
```

### Operaciones comunes
- insertar → O(1)
- buscar → O(1)
- eliminar → O(1)

Ejemplo:

```python
tabla["ciudad"] = "Buenos Aires"
del tabla["edad"]
tabla
```

Las tablas hash son fundamentales en:
- bases de datos
- caches
- indexación
- machine learning (mapeos)
- compiladores

Son una de las estructuras más importantes de la informática moderna.

Las tuplas son estructuras de datos fundamentales en Python. Aunque se parecen a las listas, tienen una característica clave: **son inmutables**. Esto significa que, una vez creadas, sus elementos no pueden modificarse.

- Métodos disponibles
- Tuplas anidadas
- Desempaquetado

## 8.1 ¿Qué es una tupla?
Una tupla es una colección **ordenada** e **inmutable** de elementos. Se define usando paréntesis:

```python
tupla = (1, 2, 3)
```

Las tuplas pueden contener elementos de distintos tipos, igual que las listas.

Ventajas de las tuplas:
- Son más rápidas que las listas
- Ocupan menos memoria
- Son ideales para datos que no deben cambiar
- Se pueden usar como claves en diccionarios (las listas no)


````python
tupla = (10, "UNO", True, 3.14)
tupla
````

## 8.2 Indexación
Las tuplas permiten acceder a sus elementos mediante índices, igual que las listas.

Ejemplo:
```
(10, "UNO", True, 3.14)
 0     1      2     3
```


````python
tupla = (10, "UNO", True, 3.14)
tupla[0], tupla[1], tupla[3]
````

## 8.3 Indexación negativa
También se puede acceder desde el final usando índices negativos.

```
(10, "UNO", True, 3.14)
 -4    -3    -2    -1
```


````python
tupla[-1], tupla[-2], tupla[-4]
````

## 8.4 Slicing (subtuplas)
El slicing funciona igual que en listas.

Sintaxis:
```
tupla[inicio:fin]
```

Ejemplos:


````python
t = (10, 20, 30, 40, 50)
t[1:4], t[:3], t[2:], t[:]
````

## 8.5 Inmutabilidad
La característica principal de las tuplas es que **no se pueden modificar**.

Intentar cambiar un elemento produce un error.


````python
t = (1, 2, 3)
try:
    t[1] = 99
except TypeError as e:
    print("Error:", e)
````

## 8.6 Métodos de tuplas
Las tuplas tienen pocos métodos debido a su inmutabilidad.

Métodos disponibles:
- `count(valor)` → cuenta cuántas veces aparece un valor
- `index(valor)` → devuelve la posición de un valor


````python
t = (1, 2, 2, 3, 2)
t.count(2), t.index(3)
````

## 8.7 Tuplas anidadas
Una tupla puede contener otras tuplas.

Esto permite representar estructuras como matrices.


````python
matriz = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)
matriz
````

````python
matriz[1][2]
````

## 8.8 Desempaquetado de tuplas
Python permite asignar los elementos de una tupla a variables individuales.

Ejemplo clásico:


````python
persona = ("Jon", 39, "Argentina")
nombre, edad, pais = persona
nombre, edad, pais
````

También se puede usar el operador `*` para capturar múltiples valores.


````python
t = (1, 2, 3, 4, 5)
a, b, *resto = t
a, b, resto
````

## 8.9 ¿Cuándo usar tuplas en lugar de listas?
Las tuplas son preferibles cuando:
- Los datos no deben cambiar
- Se necesita mayor eficiencia
- Se requiere usar la estructura como clave en un diccionario
- Se quiere asegurar integridad de datos

Ejemplos típicos:
- Coordenadas (x, y)
- Fechas (día, mes, año)
- Registros fijos

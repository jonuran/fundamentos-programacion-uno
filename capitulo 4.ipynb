Las estructuras condicionales permiten que un programa tome decisiones.

## 4.1 La sentencia `if`
La estructura condicional más básica es `if`. Permite ejecutar un bloque de código solo si una condición es verdadera.

Sintaxis:
```python
if condicion:
    bloque
```

La condición debe ser una expresión que devuelva `True` o `False`.
El bloque debe estar indentado con **4 espacios**.


````python
edad = 20
if edad >= 18:
    print("Es mayor de edad")
````

## 4.2 Estructura `if-else`
Permite ejecutar un bloque cuando la condición es verdadera y otro cuando es falsa.

Sintaxis:
```python
if condicion:
    bloque_verdadero
else:
    bloque_falso
```


````python
nota = 6
if nota >= 7:
    print("Aprobado")
else:
    print("Desaprobado")
````

## 4.3 Estructura `if-elif-else`
Se utiliza cuando hay múltiples condiciones posibles.

Sintaxis:
```python
if condicion1:
    bloque1
elif condicion2:
    bloque2
else:
    bloque_final
```

Python evalúa las condiciones en orden y ejecuta solo el primer bloque verdadero.


````python
temperatura = 30

if temperatura < 10:
    print("Hace frío")
elif temperatura < 25:
    print("Clima agradable")
else:
    print("Hace calor")
````

## 4.4 Condiciones múltiples
Podemos combinar condiciones usando operadores lógicos:
- `and`
- `or`
- `not`

Ejemplo con `and`:


````python
edad = 25
tiene_dni = True

if edad >= 18 and tiene_dni:
    print("Puede votar")
````

Ejemplo con `or`:


````python
dia = "sábado"

if dia == "sábado" or dia == "domingo":
    print("Es fin de semana")
````

Ejemplo con `not`:


````python
llueve = False

if not llueve:
    print("Podemos salir a caminar")
````

## 4.5 Buenas prácticas
Para escribir condicionales claros y mantenibles:

- Usar nombres de variables descriptivos
- Evitar condiciones demasiado largas
- Usar paréntesis para mejorar la legibilidad
- Evitar anidar muchos `if` cuando se puede usar `elif`
- Mantener la indentación consistente

Ejemplo de mala práctica:
```python
if x>10 and y<5 or z==3:
    print("ok")
```

Ejemplo mejorado:
```python
if (x > 10 and y < 5) or z == 3:
    print("ok")
```

## 24.1 Estilo de código
El estilo de código es fundamental para que el software sea **legible**, **mantenible** y **profesional**.

Python tiene una guía oficial de estilo llamada **PEP 8**.

### ✔ Reglas principales de PEP 8

#### 1. Nombres descriptivos
- Variables: `total_ventas`, `contador`
- Funciones: `calcular_promedio()`
- Clases: `ClientePremium`

#### 2. Espacios y sangría
- Usar **4 espacios** por nivel de indentación
- Espacios alrededor de operadores:
  ```python
  x = a + b
  ```

#### 3. Líneas cortas
- Máximo recomendado: **79 caracteres**

#### 4. Comentarios claros y útiles
- No describir lo obvio
- Explicar *por qué*, no *qué*

#### 5. Código limpio
- Evitar duplicación
- Evitar funciones demasiado largas
- Evitar nombres crípticos

### Ejemplo de buen estilo:

````python
def calcular_promedio(numeros):
    """Devuelve el promedio de una lista de números."""
    if not numeros:
        return 0
    return sum(numeros) / len(numeros)

calcular_promedio([10, 20, 30])
````

Seguir un estilo consistente mejora la calidad del código y facilita el trabajo en equipo.

## 24.2 Documentación
La documentación es esencial para que otros (y vos mismo en el futuro) puedan entender el código.

### ✔ Tipos de documentación
- **Docstrings** → dentro del código
- **Comentarios** → aclaraciones puntuales
- **README** → descripción general del proyecto
- **Manual de usuario** → instrucciones de uso
- **Documentación técnica** → arquitectura, decisiones, APIs

### ✔ Docstrings (PEP 257)
Ejemplo recomendado:

````python
def convertir_celsius_a_fahrenheit(c):
    """
    Convierte grados Celsius a Fahrenheit.

    Parámetros:
        c (float): temperatura en Celsius.

    Retorna:
        float: temperatura en Fahrenheit.
    """
    return c * 9/5 + 32

convertir_celsius_a_fahrenheit(25)
````

### ✔ Comentarios útiles
Los comentarios deben explicar **intención**, no repetir el código.

Ejemplo incorrecto:
```python
i = i + 1  # sumo 1
```

Ejemplo correcto:
```python
i = i + 1  # avanzo al siguiente índice
```

## 24.3 Pruebas
Las pruebas (tests) permiten verificar que el código funciona correctamente.

### ✔ Tipos de pruebas
- **Pruebas unitarias** → prueban funciones individuales
- **Pruebas de integración** → prueban módulos completos
- **Pruebas funcionales** → prueban el sistema desde el punto de vista del usuario

### ✔ ¿Por qué hacer pruebas?
- Detectan errores temprano
- Evitan regresiones
- Mejoran la confianza en el código
- Facilitan refactorizaciones

### ✔ Pruebas unitarias con `assert` (simple)

````python
def sumar(a, b):
    return a + b

assert sumar(2, 3) == 5
assert sumar(-1, 1) == 0

"Pruebas completadas"
````

### ✔ Pruebas con `unittest` (más profesional)

````python
import unittest

class TestFunciones(unittest.TestCase):
    def test_sumar(self):
        self.assertEqual(sumar(2, 3), 5)
        self.assertEqual(sumar(-1, 1), 0)

unittest.main(argv=[''], exit=False)
````

Las pruebas son una parte esencial del desarrollo profesional.

## 19.1 DFS — Búsqueda en profundidad
DFS (*Depth-First Search*) recorre un grafo **profundizando** lo más posible antes de retroceder.

### Características
- Usa una **pila** (stack) → puede ser recursiva o iterativa
- Explora caminos largos primero
- Útil para:
  - detectar ciclos
  - encontrar componentes conectados
  - recorridos topológicos
  - resolver laberintos

### Representación visual (texto)
```
A - B - D
 \   \
  \   C - E
```

### Implementación con lista de adyacencia

````python
grafo = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "E"],
    "D": ["B"],
    "E": ["C"]
}

visitados = set()

def dfs(nodo):
    if nodo not in visitados:
        print(nodo, end=" ")
        visitados.add(nodo)
        for vecino in grafo[nodo]:
            dfs(vecino)

dfs("A")
````

### DFS iterativo (con pila explícita)

````python
def dfs_iterativo(inicio):
    visitados = set()
    pila = [inicio]

    while pila:
        nodo = pila.pop()
        if nodo not in visitados:
            print(nodo, end=" ")
            visitados.add(nodo)
            pila.extend(grafo[nodo])

dfs_iterativo("A")
````

DFS es muy eficiente:
- Complejidad temporal: **O(V + E)**
- Complejidad espacial: **O(V)**

Donde:
- V = número de vértices
- E = número de aristas


## 19.2 BFS — Búsqueda en anchura
BFS (*Breadth-First Search*) recorre un grafo **nivel por nivel**.

### Características
- Usa una **cola** (FIFO)
- Explora primero los nodos más cercanos
- Útil para:
  - encontrar el camino más corto en grafos no ponderados
  - análisis de niveles
  - propagación (virus, mensajes, señales)
  - redes sociales (distancia entre personas)

### Implementación con `deque`

````python
from collections import deque

def bfs(inicio):
    visitados = set([inicio])
    cola = deque([inicio])

    while cola:
        nodo = cola.popleft()
        print(nodo, end=" ")
        for vecino in grafo[nodo]:
            if vecino not in visitados:
                visitados.add(vecino)
                cola.append(vecino)

bfs("A")
````

BFS también tiene complejidad:
- Temporal: **O(V + E)**
- Espacial: **O(V)**

Es ideal para encontrar caminos mínimos en grafos sin pesos.

## 19.3 Dijkstra — Camino mínimo
El algoritmo de **Dijkstra** encuentra el **camino más corto** desde un nodo origen hacia todos los demás en un grafo **ponderado con pesos positivos**.

### Aplicaciones
- GPS y navegación
- redes de transporte
- optimización de rutas
- análisis de redes
- algoritmos de IA

### Idea general
1. Inicializar todas las distancias en infinito
2. La distancia al nodo inicial es 0
3. Seleccionar el nodo no visitado con menor distancia
4. Actualizar distancias a sus vecinos
5. Repetir hasta visitar todos los nodos

### Representación del grafo ponderado

````python
grafo_p = {
    "A": {"B": 4, "C": 2},
    "B": {"C": 5, "D": 10},
    "C": {"E": 3},
    "D": {"F": 11},
    "E": {"D": 4},
    "F": {}
}
grafo_p
````

### Implementación de Dijkstra (versión simple)

````python
import math

def dijkstra(grafo, inicio):
    dist = {nodo: math.inf for nodo in grafo}
    dist[inicio] = 0
    visitados = set()

    while len(visitados) < len(grafo):
        nodo_actual = min((n for n in grafo if n not in visitados), key=lambda n: dist[n])
        visitados.add(nodo_actual)

        for vecino, peso in grafo[nodo_actual].items():
            nueva_dist = dist[nodo_actual] + peso
            if nueva_dist < dist[vecino]:
                dist[vecino] = nueva_dist

    return dist

dijkstra(grafo_p, "A")
````

### Complejidad
- Versión simple: **O(V²)**
- Con cola de prioridad (heap): **O((V + E) log V)**

Dijkstra es uno de los algoritmos más importantes de la informática moderna.

## 9.1 ¿Qué es un diccionario?
Un diccionario es una colección **no ordenada**, **mutable** y **indexada por claves**.

Se define usando llaves `{}`:
```python
dic = {"clave": "valor"}
```

Características principales:
- Las claves deben ser únicas
- Las claves deben ser inmutables (str, int, float, tuplas)
- Los valores pueden ser de cualquier tipo
- Permiten acceso rápido por clave

Ejemplo básico:


````python
persona = {"nombre": "Jon", "edad": 39, "pais": "Argentina"}
persona
````

## 9.2 Acceso a valores
Para acceder a un valor se usa su clave entre corchetes.

```python
dic["clave"]
```

Ejemplo:


````python
persona["nombre"], persona["edad"]
````

### Acceso seguro con `get()`
Si la clave no existe, `dic[clave]` produce error.
`get()` evita ese error devolviendo `None` o un valor por defecto.


````python
persona.get("altura"), persona.get("altura", "No disponible")
````

## 9.3 Agregar y modificar elementos
Los diccionarios son mutables, por lo que podemos agregar o modificar valores fácilmente.

### Agregar clave–valor
```python
dic["nueva_clave"] = valor
```

### Modificar clave existente
```python
dic["clave"] = nuevo_valor
```


````python
persona["altura"] = 1.80
persona["edad"] = 40
persona
````

## 9.4 Eliminar elementos
Python ofrece varias formas de eliminar elementos.

### `del dic[clave]`
Elimina una clave específica.

### `pop(clave)`
Elimina y devuelve el valor.

### `clear()`
Vacía el diccionario.


````python
persona = {"nombre": "Jon", "edad": 39, "pais": "Argentina"}
valor = persona.pop("pais")
valor, persona
````

## 9.5 Recorrer diccionarios
Podemos recorrer:
- Solo claves
- Solo valores
- Claves y valores

### Recorrer claves (por defecto)


````python
for clave in persona:
    print(clave)
````

### Recorrer valores


````python
for valor in persona.values():
    print(valor)
````

### Recorrer claves y valores


````python
for clave, valor in persona.items():
    print(clave, "=", valor)
````

## 9.6 Métodos útiles de diccionarios
Los diccionarios tienen métodos muy potentes.

### `keys()`
Devuelve todas las claves.

### `values()`
Devuelve todos los valores.

### `items()`
Devuelve pares clave–valor.

### `update()`
Actualiza el diccionario con otro diccionario.


````python
persona.update({"ciudad": "Buenos Aires", "edad": 41})
persona
````

## 9.7 Diccionarios anidados
Un diccionario puede contener otros diccionarios.

Esto es muy útil para representar estructuras complejas.


````python
alumno = {
    "nombre": "Ana",
    "cursos": {
        "Python": 9,
        "Bases de Datos": 8
    }
}
alumno
````

````python
alumno["cursos"]["Python"]
````

## 9.8 Uso práctico en ciencia de datos
Los diccionarios son esenciales en ciencia de datos porque permiten:
- Representar registros
- Convertir datos a JSON
- Construir estructuras clave–valor
- Preparar datos para Pandas
- Mapear categorías

Ejemplo: lista de diccionarios como dataset simple.


````python
dataset = [
    {"nombre": "Ana", "edad": 22, "nota": 9},
    {"nombre": "Luis", "edad": 25, "nota": 7},
    {"nombre": "Mia", "edad": 21, "nota": 8}
]
dataset
````

````python
[alumno["nota"] for alumno in dataset]
````

## 14.1 Objetos
Un **objeto** es una entidad que combina:
- **Atributos** → datos
- **Métodos** → funciones asociadas

En Python, **todo es un objeto**:
- números
- strings
- listas
- funciones
- módulos

Ejemplo:

````python
x = 10
type(x)
````

Incluso las funciones son objetos:

````python
def saludar():
    print("Hola")

type(saludar)
````

## 14.2 Clases
Una **clase** es un molde o plantilla para crear objetos.

Sintaxis básica:
```python
class Nombre:
    instrucciones
```

Ejemplo de clase vacía:

````python
class Persona:
    pass

p = Persona()
p
````

Una clase define:
- cómo se crean los objetos
- qué atributos tienen
- qué métodos pueden ejecutar

Los objetos creados a partir de una clase se llaman **instancias**.

## 14.3 Atributos y métodos
Los **atributos** son variables dentro de un objeto.
Los **métodos** son funciones dentro de una clase.

Ejemplo:

````python
class Persona:
    def saludar(self):
        return "Hola, soy una persona"

p = Persona()
p.saludar()
````

El parámetro `self` representa a la instancia actual.

Python lo pasa automáticamente.

## 14.4 Método `__init__`
El método `__init__` es el **constructor** de la clase.
Se ejecuta automáticamente al crear un objeto.

Ejemplo:

````python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

p = Persona("Ana", 22)
p.nombre, p.edad
````

El constructor permite inicializar atributos obligatorios o valores por defecto.

## 14.5 Método `__str__`
El método `__str__` define cómo se muestra un objeto como texto.

Ejemplo:

````python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    def __str__(self):
        return f"Persona: {self.nombre}"

p = Persona("Jonatan")
print(p)
````

`__str__` mejora la legibilidad y es muy útil para debugging.

## 14.6 Atributos de clase vs instancia
### ✔ Atributos de instancia
Son propios de cada objeto.

### ✔ Atributos de clase
Son compartidos por **todas** las instancias.

Ejemplo:

````python
class Persona:
    especie = "Humano"  # atributo de clase

    def __init__(self, nombre):
        self.nombre = nombre  # atributo de instancia

p1 = Persona("Ana")
p2 = Persona("Luis")

p1.especie, p2.especie
````

Si cambiamos el atributo de clase, afecta a todas las instancias.

````python
Persona.especie = "Ser humano"
p1.especie, p2.especie
````

## 14.7 Herencia
La herencia permite crear clases basadas en otras.

Ventajas:
- reutilización de código
- extensibilidad
- organización jerárquica

Ejemplo básico:

````python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    def saludar(self):
        return f"Hola, soy {self.nombre}"

class Estudiante(Persona):
    def estudiar(self):
        return f"{self.nombre} está estudiando"

e = Estudiante("Mia")
e.saludar(), e.estudiar()
````

### Sobrescritura de métodos
Una subclase puede redefinir métodos de la clase padre.

````python
class Estudiante(Persona):
    def saludar(self):
        return f"Hola, soy {self.nombre} y soy estudiante"

Estudiante("Ana").saludar()
````

## 14.8 Polimorfismo
El polimorfismo permite que distintas clases implementen métodos con el mismo nombre.

Esto permite escribir código más flexible.

Ejemplo clásico:

````python
class Perro:
    def hablar(self):
        return "Guau"

class Gato:
    def hablar(self):
        return "Miau"

animales = [Perro(), Gato()]
[a.hablar() for a in animales]
````

El polimorfismo es clave en:
- frameworks
- librerías
- diseño de APIs
- patrones de diseño

Permite escribir funciones que aceptan múltiples tipos de objetos mientras compartan un mismo método.

## 15.1 Qué es un módulo
Un **módulo** es simplemente un archivo `.py` que contiene código.

Ejemplo de módulo llamado `operaciones.py`:
```python
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b
```

Luego puede importarse desde otro archivo.

### ¿Por qué usar módulos?
- Organización del código
- Reutilización
- Separación de responsabilidades
- Facilitar pruebas unitarias
- Evitar duplicación

En proyectos reales, cada archivo suele representar un módulo con una responsabilidad clara.

## 15.2 `import`
La forma más básica de usar un módulo es con `import`.

Ejemplo:

````python
import math
math.sqrt(25), math.pi
````

Cuando importamos un módulo:
- Python lo busca en el **path de módulos**
- Lo carga en memoria
- Lo ejecuta una sola vez

Podemos asignar un alias para abreviar:

````python
import math as m
m.cos(0)
````

### ¿Dónde busca Python los módulos?
En el orden:
1. Directorio actual
2. Variables de entorno (`PYTHONPATH`)
3. Librería estándar
4. Librerías instaladas con pip

Podemos ver el path:

````python
import sys
sys.path
````

## 15.3 `from import`
Permite importar partes específicas de un módulo.

Ejemplo:

````python
from math import sqrt, pi
sqrt(16), pi
````

### Importar todo (no recomendado)
```python
from math import *
```
Problemas:
- Contamina el espacio de nombres
- Puede sobrescribir funciones
- Dificulta la lectura del código

Es mejor importar solo lo necesario o usar alias.

## 15.4 Librería estándar
Python incluye una **librería estándar** enorme, lista para usar sin instalar nada.

Algunos módulos importantes:
- `math` → funciones matemáticas
- `random` → generación de números aleatorios
- `datetime` → fechas y tiempos
- `os` → interacción con el sistema operativo
- `sys` → información del intérprete
- `json` → trabajar con JSON
- `csv` → archivos CSV
- `statistics` → estadísticas básicas

Ejemplo con `random`:

````python
import random
random.randint(1, 10)
````

Ejemplo con `datetime`:

````python
from datetime import datetime
datetime.now()
````

La librería estándar es uno de los puntos fuertes de Python: permite resolver problemas comunes sin instalar nada adicional.

## 15.5 Librerías externas
Además de la librería estándar, Python permite instalar librerías externas mediante **pip**.

Ejemplos populares:
- `numpy` → cálculo numérico
- `pandas` → análisis de datos
- `matplotlib` → gráficos
- `requests` → HTTP
- `scikit-learn` → machine learning
- `flask` / `django` → desarrollo web

Instalación (en terminal):
```bash
pip install numpy
```

Uso:

````python
import numpy as np
np.array([1, 2, 3])
````

### ¿Dónde se instalan las librerías externas?
En el directorio `site-packages`, dentro de la instalación de Python.

Podemos ver todas las librerías instaladas con:
```bash
pip list
```

O desde Python:

````python
import pkgutil
[mod.name for mod in pkgutil.iter_modules()][:20]  # primeros 20 módulos
````

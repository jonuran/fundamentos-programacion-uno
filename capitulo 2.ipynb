## 2.1 Tipos de datos primitivos
Los tipos primitivos representan valores simples. Python maneja automáticamente su creación y gestión.

Los principales son:
- `int`: números enteros
- `float`: números reales
- `str`: cadenas de texto
- `bool`: valores lógicos

### 2.1.1 Números (`int`, `float`)
Los enteros pueden ser positivos, negativos o cero. Los flotantes representan números con decimales.


```python
entero = 42
decimal = 3.14159
negativo = -12
entero, decimal, negativo
```

### 2.1.2 Cadenas (`str`)
Una cadena es una secuencia de caracteres. Se puede escribir con comillas simples o dobles.
Python permite cadenas vacías, cadenas con espacios y caracteres especiales como saltos de línea.


```python
cadena1 = "Hola UNO"
cadena2 = 'Python es genial'
cadena_vacia = ""
cadena1, cadena2, cadena_vacia
```

### 2.1.3 Booleanos (`bool`)
Representan valores lógicos: `True` o `False`.
Se utilizan en condiciones, comparaciones y estructuras de control.


```python
activo = True
aprobado = False
activo, aprobado
```

## 2.2 Tipos de datos compuestos
Los tipos compuestos permiten almacenar colecciones de valores.

### 2.2.1 Listas
Son colecciones ordenadas y mutables. Pueden contener elementos de distintos tipos.


```python
lista = [1, "dos", True, 3.14]
lista
```

### 2.2.2 Tuplas
Son colecciones ordenadas pero **inmutables**. Una vez creadas, no pueden modificarse.
Son útiles para representar datos que no deben cambiar.


```python
tupla = (10, 20, 30)
tupla
```

### 2.2.3 Diccionarios
Son colecciones de pares **clave–valor**. Las claves deben ser únicas.
Son ideales para representar datos estructurados.


```python
dic = {"nombre": "UNO", "anio": 2024, "activo": True}
dic
```

## 2.3 type(): conocer el tipo de un dato
La función `type()` devuelve la clase del objeto.


```python
type(1), type(3.14), type("Hola"), type([1,2]), type({"a":1})
```

## 2.4 Conversión de tipos
Python permite convertir entre tipos siempre que la conversión tenga sentido.

- `int()` convierte a entero
- `float()` convierte a real
- `str()` convierte a cadena
- `bool()` convierte a booleano


```python
int("12"), float("3.14"), str(3.14), bool(""), bool("Python")
```

## 2.5 Variables
Una variable es un nombre que referencia un valor en memoria.

Reglas:
- Deben comenzar con letra o guion bajo
- No pueden contener espacios
- No pueden ser palabras reservadas
- Python usa **tipado dinámico**

Asignación múltiple e intercambio son características muy útiles.


```python
a, b = 1, 2
a, b = b, a
a, b
```

## 2.7 Salida de datos (print)
La función `print()` permite mostrar información en pantalla.
Parámetros útiles:
- `sep`: separador
- `end`: finalización


```python
print("Hola", "UNO", sep="-")
```

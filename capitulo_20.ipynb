## 20.1 Qué es la complejidad
La complejidad computacional mide **cuántos recursos** necesita un algoritmo para ejecutarse.

Los dos recursos principales son:
- **Tiempo** → cuántas operaciones realiza
- **Espacio** → cuánta memoria utiliza

La complejidad se expresa usando **notación Big O**, que veremos en el capítulo siguiente.

### ¿Por qué es importante?
- Permite comparar algoritmos
- Ayuda a predecir rendimiento
- Evita soluciones ineficientes
- Es clave para trabajar con grandes datasets

### Ejemplo intuitivo
Supongamos que queremos buscar un número en una lista:
- Búsqueda lineal → revisa uno por uno
- Búsqueda binaria → divide la lista a la mitad cada vez

Claramente, la búsqueda binaria es mucho más eficiente.

La complejidad nos permite **cuantificar** esa diferencia.

## 20.2 Complejidad temporal
La **complejidad temporal** mide cuántas operaciones realiza un algoritmo en función del tamaño de entrada `n`.

### Ejemplo 1: bucle simple
```python
for i in range(n):
    print(i)
```
Realiza `n` operaciones → **O(n)**.

### Ejemplo 2: doble bucle
```python
for i in range(n):
    for j in range(n):
        print(i, j)
```
Realiza `n × n` operaciones → **O(n²)**.

### Ejemplo 3: acceso directo
```python
lista[10]
```
Acceso inmediato → **O(1)**.

### Ejemplo práctico en Python

````python
def suma_lineal(n):
    total = 0
    for i in range(n):
        total += i
    return total

suma_lineal(10)
````

Este algoritmo es **O(n)** porque el tiempo crece proporcionalmente al tamaño de `n`.

### Comparación conceptual
- O(1) → constante
- O(log n) → muy eficiente
- O(n) → lineal
- O(n log n) → típico de algoritmos de ordenamiento
- O(n²) → cuadrático, lento para grandes n
- O(2ⁿ) → explosivo, impráctico
- O(n!) → extremadamente lento

La complejidad temporal es clave para elegir algoritmos adecuados.

## 20.3 Complejidad espacial
La **complejidad espacial** mide cuánta memoria adicional necesita un algoritmo.

### Ejemplo 1: sin memoria extra
```python
def suma(lista):
    total = 0
    for x in lista:
        total += x
    return total
```
Usa solo una variable → **O(1)**.

### Ejemplo 2: crea una lista nueva
```python
def duplicar(lista):
    nueva = []
    for x in lista:
        nueva.append(x * 2)
    return nueva
```
Crea una lista del mismo tamaño → **O(n)**.

### Ejemplo práctico

````python
def crear_lista(n):
    return [i for i in range(n)]

crear_lista(10)
````

Este algoritmo usa memoria proporcional a `n` → **O(n)**.

### Importancia en ciencia de datos
Cuando trabajamos con:
- millones de filas
- imágenes grandes
- modelos de machine learning

La memoria puede ser un recurso crítico.

Por eso es importante elegir algoritmos que no consuman memoria innecesaria.

## 20.4 Medición de tiempo
Python permite medir el tiempo de ejecución de un algoritmo.

### Método 1: `time.time()`
Mide tiempo en segundos.

### Método 2: `time.perf_counter()`
Más preciso para benchmarking.

### Ejemplo práctico

````python
import time

def prueba(n):
    total = 0
    for i in range(n):
        total += i
    return total

inicio = time.perf_counter()
prueba(1_000_000)
fin = time.perf_counter()

fin - inicio
````

Podemos comparar tiempos de diferentes algoritmos.

### Ejemplo: comparación entre dos funciones

````python
def suma_for(n):
    total = 0
    for i in range(n):
        total += i
    return total

def suma_formula(n):
    return n * (n - 1) // 2

for f in [suma_for, suma_formula]:
    inicio = time.perf_counter()
    f(1_000_000)
    fin = time.perf_counter()
    print(f.__name__, fin - inicio)
````

Claramente:
- `suma_for` → O(n)
- `suma_formula` → O(1)

La diferencia de rendimiento es enorme.

## 23.1 Búsqueda binaria
La **búsqueda binaria** es un algoritmo extremadamente eficiente para buscar un elemento en una lista **ordenada**.

### Idea principal
1. Tomar el elemento del medio.
2. Compararlo con el objetivo.
3. Si es mayor, buscar en la mitad izquierda.
4. Si es menor, buscar en la mitad derecha.
5. Repetir hasta encontrar el elemento o agotar la lista.

### Complejidad
- **O(log n)** → muy eficiente

### Representación visual (texto)
Lista ordenada:
```
[2, 5, 7, 10, 14, 20, 25]
```
Buscar 14:
```
1. Medio = 10 → 14 > 10 → derecha
2. Medio = 20 → 14 < 20 → izquierda
3. Medio = 14 → encontrado
```

### Implementación en Python

````python
def busqueda_binaria(lista, objetivo):
    izq, der = 0, len(lista) - 1
    while izq <= der:
        medio = (izq + der) // 2
        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            izq = medio + 1
        else:
            der = medio - 1
    return -1

lista = [2, 5, 7, 10, 14, 20, 25]
busqueda_binaria(lista, 14)
````

La búsqueda binaria es uno de los algoritmos más importantes de la informática debido a su eficiencia.

## 23.2 Ordenamiento burbuja (Bubble Sort)
Bubble Sort es uno de los algoritmos de ordenamiento más simples, aunque no es eficiente.

### Idea principal
1. Recorrer la lista comparando pares de elementos.
2. Si están en el orden incorrecto, intercambiarlos.
3. Repetir hasta que no haya más intercambios.

### Representación visual (texto)
Lista:
```
[5, 3, 8, 1]
```
Primer recorrido:
```
5 > 3 → swap → [3, 5, 8, 1]
5 < 8 → ok
8 > 1 → swap → [3, 5, 1, 8]
```

### Complejidad
- **O(n²)** → muy lento para listas grandes

### Implementación en Python

````python
def bubble_sort(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

bubble_sort([5, 3, 8, 1, 2])
````

Bubble Sort es útil para aprender, pero no se usa en producción debido a su baja eficiencia.

Sin embargo, es excelente para enseñar conceptos como:
- comparaciones
- intercambios
- bucles anidados
- análisis de complejidad


## 23.3 Comparación de eficiencia
Compararemos:
- Búsqueda lineal → O(n)
- Búsqueda binaria → O(log n)
- Bubble Sort → O(n²)

### Comparación conceptual
```
n = 1.000.000

Búsqueda lineal:     1.000.000 pasos
Búsqueda binaria:    ~20 pasos
Bubble Sort:         1.000.000.000.000 pasos
```

### Comparación práctica en Python

````python
import time

def busqueda_lineal(lista, objetivo):
    for i, x in enumerate(lista):
        if x == objetivo:
            return i
    return -1

lista = list(range(1_000_000))

inicio = time.perf_counter()
busqueda_lineal(lista, 999_999)
print("Lineal:", time.perf_counter() - inicio)

inicio = time.perf_counter()
busqueda_binaria(lista, 999_999)
print("Binaria:", time.perf_counter() - inicio)
````

La diferencia es enorme:
- La búsqueda binaria es miles de veces más rápida.
- Bubble Sort es impráctico para listas grandes.

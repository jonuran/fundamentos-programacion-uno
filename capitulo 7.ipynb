## 7.1 ¿Qué es una lista?
Una lista es una colección **ordenada** y **mutable** de elementos. Puede contener valores de distintos tipos: números, cadenas, booleanos, otras listas, etc.

Se define usando corchetes:
```python
lista = [1, 2, 3]
```

Las listas permiten:
- Acceso por índice
- Modificación de elementos
- Agregar y eliminar valores
- Recorrer sus elementos
- Ordenarlas


````python
lista = [10, "UNO", True, 3.14]
lista
````

## 7.2 Indexación
Cada elemento tiene una posición (índice), comenzando desde 0.

Ejemplo:
```
[10, "UNO", True, 3.14]
 0     1      2     3
```

Acceso por índice:


````python
lista = [10, "UNO", True, 3.14]
lista[0], lista[1], lista[3]
````

## 7.3 Indexación negativa
Permite acceder desde el final.

```
[10, "UNO", True, 3.14]
 -4    -3    -2    -1
```


````python
lista[-1], lista[-2], lista[-4]
````

## 7.4 Slicing (sublistas)
Permite obtener partes de una lista.

Sintaxis:
```
lista[inicio:fin]
```
Python incluye `inicio` pero excluye `fin`.


````python
lista = [10, 20, 30, 40, 50]
lista[1:4], lista[:3], lista[2:], lista[:]
````

## 7.5 Modificación de listas
Las listas son mutables: podemos cambiar sus elementos.


````python
lista = [1, 2, 3]
lista[1] = 99
lista
````

## 7.6 Agregar elementos
### `append()`
Agrega un elemento al final.


````python
lista = [1, 2, 3]
lista.append(4)
lista
````

### `insert()`
Inserta un elemento en una posición específica.


````python
lista.insert(1, "UNO")
lista
````

### `extend()`
Agrega múltiples elementos.


````python
lista.extend([100, 200])
lista
````

## 7.7 Eliminar elementos
### `remove()`
Elimina la primera aparición de un valor.


````python
lista = [1, 2, 3, 2]
lista.remove(2)
lista
````

### `pop()`
Elimina y devuelve un elemento por índice.


````python
lista = [10, 20, 30]
valor = lista.pop(1)
valor, lista
````

### `clear()`
Vacía la lista.


````python
lista.clear()
lista
````

## 7.8 Ordenamiento
Python permite ordenar listas numéricas o alfabéticas.

### `sort()` modifica la lista
### `sorted()` devuelve una nueva lista


````python
numeros = [5, 1, 4, 2, 3]
numeros.sort()
numeros
````

````python
sorted([5, 1, 4, 2, 3])
````

## 7.9 Listas anidadas
Una lista puede contener otras listas.
Esto permite representar matrices o estructuras complejas.


````python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
matriz
````

Acceso a elementos anidados:


````python
matriz[1][2]
````

## 7.10 List comprehension
Es una forma compacta y eficiente de crear listas.

Sintaxis:
```
[expresion for elemento in iterable]
```

Ejemplo: cuadrados del 1 al 5.


````python
[x*x for x in range(1, 6)]
````

## ¿Por qué son importantes las excepciones?
Las excepciones permiten:
- Evitar que el programa se detenga inesperadamente.
- Mostrar mensajes claros al usuario.
- Registrar errores para depuración.
- Continuar la ejecución incluso si ocurre un problema.
- Manejar datos faltantes o inválidos.

En ciencia de datos, son esenciales para:
- Validar entradas.
- Manejar archivos inexistentes.
- Controlar errores en conversiones de tipos.
- Evitar fallos en pipelines de procesamiento.


## 13.1 Qué son las excepciones
Una excepción es un **evento anómalo** que ocurre durante la ejecución del programa y que interrumpe el flujo normal.

Ejemplos comunes:
- División por cero → `ZeroDivisionError`
- Índice fuera de rango → `IndexError`
- Conversión inválida → `ValueError`
- Archivo inexistente → `FileNotFoundError`
- Acceso a variable no definida → `NameError`

Ejemplo de error sin manejar:

````python
10 / 0  # ZeroDivisionError
````

Cuando ocurre una excepción, Python detiene el programa a menos que la manejemos con `try-except`.

## 13.2 Tipos de excepciones
Python incluye una gran cantidad de excepciones predefinidas.

Algunas de las más comunes:

| Excepción | Descripción |
|----------|-------------|
| `TypeError` | Tipo incorrecto |
| `ValueError` | Valor inválido |
| `IndexError` | Índice fuera de rango |
| `KeyError` | Clave inexistente en diccionario |
| `ZeroDivisionError` | División por cero |
| `FileNotFoundError` | Archivo no encontrado |
| `IOError` | Error de entrada/salida |
| `NameError` | Variable no definida |
| `AttributeError` | Atributo inexistente |

Ejemplo de `ValueError`:

````python
int("hola")  # ValueError
````

Ejemplo de `IndexError`:

````python
lista = [1, 2, 3]
lista[10]
````

## 13.3 Manejo con try-except
La estructura básica es:
```python
try:
    código que puede fallar
except TipoDeError:
    acción si ocurre ese error
```

Ejemplo:

````python
try:
    numero = int("hola")
except ValueError:
    print("Error: no se puede convertir a entero")
````

### Capturar múltiples excepciones
Podemos manejar distintos errores por separado:

````python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("No se puede dividir por cero")
except TypeError:
    print("Tipo incorrecto")
````

También podemos capturar varias excepciones juntas:

````python
try:
    x = int(None)
except (TypeError, ValueError) as e:
    print("Error capturado:", e)
````

### Bloque `else`
Se ejecuta **solo si no hubo errores**.

````python
try:
    n = int("123")
except ValueError:
    print("Error de conversión")
else:
    print("Conversión exitosa:", n)
````

## 13.4 finally
El bloque `finally` se ejecuta **siempre**, haya o no error.

Es ideal para liberar recursos:
- cerrar archivos
- cerrar conexiones
- liberar memoria

Ejemplo:

````python
try:
    f = open("archivo_inexistente.txt")
except FileNotFoundError:
    print("Archivo no encontrado")
finally:
    print("Fin del bloque try-except-finally")
````

Otro ejemplo práctico:

````python
try:
    print(10 / 2)
finally:
    print("Esto se ejecuta siempre")
````

## 13.5 Buenas prácticas
Para manejar excepciones correctamente, se recomienda:

### ✔️ Capturar solo las excepciones necesarias
Evitar:
```python
except:
    pass
```
Esto oculta errores reales.

### ✔️ Usar mensajes claros
Ayuda a depurar y entender qué falló.

### ✔️ Usar `finally` para liberar recursos
Especialmente con archivos.

### ✔️ Validar datos antes de operar
Ejemplo:
```python
if isinstance(x, int):
    ...
```

### ✔️ No usar excepciones para control de flujo normal
Las excepciones deben ser **excepcionales**, no parte del flujo habitual.

### ✔️ Registrar errores en aplicaciones grandes
Con `logging`.

## Ejemplo completo integrando todo
Un ejemplo realista de lectura segura de archivos:

````python
def leer_archivo(ruta):
    try:
        with open(ruta, "r") as f:
            return f.read()
    except FileNotFoundError:
        print("Error: el archivo no existe")
    except PermissionError:
        print("Error: no tienes permisos para leer este archivo")
    except Exception as e:
        print("Error inesperado:", e)

leer_archivo("no_existe.txt")
````

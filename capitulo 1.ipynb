## 1.1 ¿Qué es Python?
Python es un lenguaje de programación **de alto nivel**, creado en 1991 y mantenido por la Python Software Foundation. Se caracteriza por ser:

- **Multiparadigma**: permite programación imperativa, funcional y orientada a objetos.
- **Interpretado**: no requiere compilación previa.
- **Tipado dinámico**: las variables no requieren declaración de tipo.
- **Multiplataforma**: funciona en Windows, Linux y macOS.

Su filosofía se basa en el "Zen de Python":
- Bello es mejor que feo.
- Simple es mejor que complejo.
- Legible es mejor que ilegible.

Python se utiliza en:
- Ciencia de datos
- Machine learning
- Desarrollo web
- Automatización
- Robótica
- Educación


## 1.2 Ventajas de Python
Python se destaca por:

- **Sintaxis clara y legible**
- **Gran comunidad** y abundante documentación
- **Miles de librerías** (NumPy, Pandas, Matplotlib, TensorFlow, etc.)
- **Código abierto**
- **Portabilidad**
- **Integración con otros lenguajes** (C, C++, Java)
- **Ideal para prototipado rápido**

Estas características lo convierten en un lenguaje ideal para estudiantes y profesionales.


## 1.3 Tipos de ejecución
Python puede ejecutarse de dos formas principales:

### 1.3.1 Modo interactivo (REPL)
Permite ejecutar instrucciones una por una.


````python
nombre = "UNO"
print("Hola", nombre)
````

### 1.3.2 Modo script
Se escribe un archivo `.py` y se ejecuta completo.

Ejemplo:
```python
nombre = "UNO"
print("Hola", nombre)
```

Luego en terminal:
```
python archivo.py
```


## 1.4 Consola vs Script
| Consola | Script |
|--------|--------|
| Ejecuta línea por línea | Ejecuta todo el archivo |
| Ideal para pruebas | Ideal para proyectos |
| No guarda código | Código persistente |


## 1.5 Estructura básica de un programa
Un programa típico en Python contiene:
- Importaciones
- Definición de funciones
- Código principal
- Manejo de errores (opcional)


````python
def sumar(a, b):
    return a + b

print(sumar(3, 4))
````

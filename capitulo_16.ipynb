La depuración no solo consiste en "arreglar errores", sino en **comprender el comportamiento real del programa**, compararlo con el comportamiento esperado y tomar decisiones informadas.

## 16.1 Qué es depurar
Depurar es el proceso de:
1. Detectar un error (bug)
2. Reproducirlo
3. Analizar su causa
4. Corregirlo
5. Verificar que no afecte otras partes del programa

### Tipos de errores
- **Errores de sintaxis**: el código no puede ejecutarse.
- **Errores de ejecución**: ocurren mientras el programa corre.
- **Errores lógicos**: el programa corre, pero hace algo incorrecto.

Ejemplo de error lógico (difícil de detectar):

````python
def promedio(lista):
    return sum(lista) / len(lista)

promedio([10, 20, 30])  # correcto
promedio([])  # error lógico: división por cero
````

Los errores lógicos son los más difíciles de detectar porque **no generan excepciones**, pero producen resultados incorrectos.

## 16.2 Técnicas de depuración
Existen varias técnicas para depurar código de forma eficiente.

### ✔ 1. Reproducir el error
Antes de corregir un error, es fundamental **poder reproducirlo consistentemente**.

### ✔ 2. Aislar el problema
Reducir el código al mínimo que produce el error.

### ✔ 3. Leer el mensaje de error
Python ofrece mensajes de error muy descriptivos.

Ejemplo:

````python
int("hola")  # ValueError: invalid literal for int()
````

### ✔ 4. Trazar el flujo del programa
Entender qué partes del código se ejecutan y en qué orden.

### ✔ 5. Revisar valores intermedios
Verificar que las variables contengan lo esperado.

### ✔ 6. Depurar con herramientas
Python incluye `pdb`, y VS Code tiene un depurador visual muy potente.

### ✔ 7. Pensar en casos extremos
Ejemplo: listas vacías, valores nulos, tipos inesperados.

### ✔ 8. Agregar tests
Los tests automatizados ayudan a detectar errores antes de que ocurran.

## 16.3 Uso de `print`
La técnica más simple (y sorprendentemente útil) para depurar es imprimir valores.

Ejemplo de error:

````python
def dividir(a, b):
    return a / b

dividir(10, 0)
````

Depuración con `print`:

````python
def dividir(a, b):
    print("a =", a)
    print("b =", b)
    return a / b

dividir(10, 0)
````

Aunque `print` es útil, tiene limitaciones:
- ensucia el código
- no permite inspeccionar el estado completo
- no permite pausar la ejecución

Por eso existen depuradores profesionales.

## 16.4 Depuración en VS Code
VS Code incluye un depurador visual muy potente que permite:
- ejecutar el programa paso a paso
- inspeccionar variables
- agregar puntos de interrupción (breakpoints)
- ver el call stack
- evaluar expresiones en tiempo real

### ✔ ¿Qué es un breakpoint?
Un **breakpoint** es una marca que indica dónde detener la ejecución.

Cuando el programa se detiene:
- podemos ver el valor de todas las variables
- podemos avanzar línea por línea
- podemos ejecutar expresiones manualmente

### ✔ Ejemplo de código para depurar en VS Code
```python
def calcular_total(precios):
    total = 0
    for p in precios:
        total += p
    return total

print(calcular_total([10, 20, 30]))
```

Podemos colocar un breakpoint en la línea `total += p` para ver cómo cambia el valor.

### ✔ Modos de ejecución en el depurador
- **Continue** → sigue hasta el próximo breakpoint
- **Step Over** → ejecuta la línea actual
- **Step Into** → entra dentro de funciones
- **Step Out** → sale de la función actual

### ✔ Ventanas útiles en VS Code
- **Variables** → muestra valores actuales
- **Watch** → permite monitorear expresiones
- **Call Stack** → muestra el orden de llamadas
- **Breakpoints** → lista de puntos de interrupción

### ✔ Depuración de scripts de Python
VS Code detecta automáticamente el entorno Python y permite depurar con un clic.

### ✔ Depuración interactiva
VS Code permite abrir una consola interactiva mientras el programa está detenido.

Esto permite probar expresiones, inspeccionar objetos y modificar variables en tiempo real.

## 11.1 Funciones recursivas
Una función recursiva es aquella que **se llama a sí misma**.

La recursión es útil para problemas que pueden dividirse en subproblemas más pequeños, como:
- factorial
- secuencia Fibonacci
- recorridos de árboles
- algoritmos de búsqueda

Toda función recursiva debe tener:
- **Caso base**: condición que detiene la recursión.
- **Caso recursivo**: llamada a sí misma.

Ejemplo: factorial.

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

factorial(5)
````

Otro ejemplo: Fibonacci.

````python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

fibonacci(10)
````

## 11.2 Riesgos de la recursión
La recursión es elegante, pero tiene riesgos:

### 1. **Límite de recursión**
Python tiene un límite de llamadas recursivas (generalmente 1000).

### 2. **Consumo de memoria**
Cada llamada recursiva ocupa espacio en la pila.

### 3. **Ineficiencia**
Algunas funciones recursivas recalculan valores muchas veces (como Fibonacci).

### Alternativa: recursión optimizada o iteración
Ejemplo iterativo de Fibonacci:

````python
def fibonacci_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

fibonacci_iter(10)
````

## 11.3 Funciones lambda
Las funciones `lambda` son funciones **anónimas**, pequeñas y de una sola expresión.

Sintaxis:
```python
lambda argumentos: expresión
```

Ejemplo:

````python
cuadrado = lambda x: x * x
cuadrado(6)
````

Las lambdas se usan mucho con funciones como `map`, `filter`, `sorted`, etc.

## 11.4 `map()`
`map()` aplica una función a cada elemento de un iterable.

Sintaxis:
```python
map(funcion, iterable)
```

Ejemplo: convertir lista de strings a enteros.

````python
numeros = ["1", "2", "3", "4"]
list(map(int, numeros))
````

Ejemplo con lambda:

````python
list(map(lambda x: x * 2, [1, 2, 3, 4]))
````

## 11.5 `filter()`
`filter()` selecciona elementos que cumplen una condición.

Sintaxis:
```python
filter(funcion, iterable)
```

Ejemplo: filtrar números pares.

````python
list(filter(lambda x: x % 2 == 0, range(10)))
````

## 11.6 `zip()`
`zip()` combina elementos de múltiples iterables.

Ejemplo:

````python
nombres = ["Ana", "Luis", "Mia"]
notas = [9, 7, 8]

list(zip(nombres, notas))
````

También se puede usar para transponer matrices:

````python
matriz = [[1, 2, 3], [4, 5, 6]]
list(zip(*matriz))
````

## 11.7 `reduce()`
`reduce()` aplica una función acumulativa a los elementos de un iterable.

No está en el core de Python: se importa desde `functools`.

Ejemplo: sumar todos los elementos.

````python
from functools import reduce

reduce(lambda a, b: a + b, [1, 2, 3, 4, 5])
````

Otro ejemplo: producto total.

````python
reduce(lambda a, b: a * b, [1, 2, 3, 4])
````

## 11.8 Comprensión de listas
Las *list comprehensions* permiten crear listas de forma compacta y eficiente.

Sintaxis:
```python
[expresion for elemento in iterable]
```

Ejemplo: cuadrados del 1 al 10.

````python
[x*x for x in range(1, 11)]
````

Con condición:

````python
[x for x in range(20) if x % 2 == 0]
````

Anidada:

````python
[(i, j) for i in range(3) for j in range(3)]
````

## 11.9 Comprensión de diccionarios
Permite crear diccionarios de forma compacta.

Sintaxis:
```python
{clave: valor for elemento in iterable}
```

Ejemplo: números y sus cuadrados.

````python
{x: x*x for x in range(1, 6)}
````

Invertir claves y valores:

````python
d = {"a": 1, "b": 2, "c": 3}
{v: k for k, v in d.items()}
````
